```python
ds = pd.read_csv('diabetes.csv')
Y = ds[['Outcome']].to_numpy().T
X = ds.drop(['Outcome'], axis=1).to_numpy().T
ds.head()
```

Paso 2. Crea una neurona logística que aprenda el dataset tratando de predecir la salida.

```python
class Logistic_Neuron:

    def __init__(self, n_inputs, learning_rate=0.1):
        self.w = - 1 + 2 * np.random.rand(n_inputs)
        self.b = - 1 + 2 * np.random.rand()
        self.eta = learning_rate

    def predict_proba(self, X):
        Z = np.dot(self.w, X) + self.b
        Y_est = 1/(1+np.exp(-Z))
        return Y_est

    def predict(self, X, umbral=0.5):
        Z = np.dot(self.w, X) + self.b
        Y_est = 1/(1+np.exp(-Z))
        return 1 * (Y_est > umbral)

    def train(self, X, Y, epochs=100):
        p = X.shape[1]
        for _ in range(epochs):
            Y_est = self.predict_proba(X)
            self.w += (self.eta/p) * np.dot((Y - Y_est), X.T).ravel()
            self.b += (self.eta/p) * np.sum(Y - Y_est)
```

Paso 3. Con un umbral de 0.5, calcula el porcentaje de los diagnósticos acertados por la neurona.

```python
from re import I
neuron = Logistic_Neuron(X.shape[0], learning_rate = 0.005)
neuron.train(X, Y, epochs=10000)

Y_predict = neuron.predict(X, 0.5).ravel()
Y = Y.ravel()

ok = 0
for i in range(Y.shape[0]):
  if Y[i] == Y_predict[i]:
    ok += 1

print(f'Porcentaje de aciertos: {100 * ok/Y.shape[0]}',"\b%")
```
